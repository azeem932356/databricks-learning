```python
from pyspark.sql import functions as F, types as T
```

```python
#creating a dataframe

rows_customers= [
    (1,"John","USA", True),
    (2,"Mary","IN", False),
    (3,"Bob","Mexico", True),
    (4,"Alice","USA", None),
    (5,"Carlos","Brazil", False),
    (6,"Sophia","USA", True)
]


rows_orders= [
    (101, 1, 120.0, "IN"), 
    (102, 1, 80.0, "USA"), 
    (103, 2, 150.0, "USA"), 
    (104, 3, 90.0, "Mexico"), 
    (105, 4, 110.0, "USA"), 
    (106, 5, 70.0, "Brazil"), 
    (107, 6, 130.0, "USA")
]
```

```python
#Creating Schema

schema_customers =T.StructType([
    T.StructField("customer_id", T.IntegerType(), True),
    T.StructField("name", T.StringType(), True),
    T.StructField("country", T.StringType(), True),
    T.StructField("VIP", T.BooleanType(), True)
])

schema_orders = T.StructType([
    T.StructField("order_id", T.IntegerType(), True),
    T.StructField("customer_id", T.IntegerType(), True),
    T.StructField("amount", T.DoubleType(), True),
    T.StructField("country", T.StringType(), True)
])


```

```python
#Passing schems

df_customers= spark.createDataFrame(rows_customers, schema_customers)
display(df_customers)
```

```python
df_orders= spark.createDataFrame(rows_orders, schema_orders)
display(df_orders)
```

### joins

```python
#inner join

df_inner= df_orders.join(df_customers, on= "customer_id", how= "inner")
display(df_inner)
```

```python
#left join

df_inner= df_orders.join(df_customers, on= "customer_id", how= "left")
display(df_inner)
```

```python
#right

df_inner= df_orders.join(df_customers, on= "customer_id", how= "right")
display(df_inner)
```

```python
#alias

o= df_orders.alias("o")
o.show()
```

```python
o, c= df_orders.alias("o"), df_customers.alias("c")
c.show()
```

```python
o.show()
```

```python
#instead of useing full name we can use alias as referece to the tables

df_inner= o.join(c, on= "customer_id", how= "inner")
df_inner.show()
```

```python
#using alias we can going to change column names for better understanding 

df_inner_clean= (
    o.join(c, on="customer_id", how= "inner")            #performing join
    .select("order_id", "customer_id", "amount",         #using-->  .select()  to select only columns we want 
    F.col("o.country").alias ("ship_country"),           #using --> alias and F to change column names 
    "name", 
    F.col("c.country").alias("cust_country"),            #using --> alias and F to change column names 
    "VIP")
)
display(df_inner_clean)
```

```python
#full join    ( returns all records from both tables, regardless of whether they have a matching row in the other table)

df_full= o.join(c, on= "customer_id", how= "full")
display(df_full)
```

```python
#outer join      ( a method used to combine rows from two or more tables such that the result includes both matching and non-matching records)

df_outer= o.join(c, on= "customer_id", how= "outer")
display(df_outer)
```

```python
#cross join     ( a type of join that combines every row from the first table with every row from the second table)

df_cross= o.join(c, on= "customer_id", how="cross")
display(df_cross)
```

```python
#semi join   ( a type of join that returns rows from the first (left) table if a matching record exists in the second (right) table, without returning any data or columns from the second table )

df_semi= o.join(c, on= "customer_id", how= "semi")
display(df_semi)
```

```python
#anti join      ( a query pattern used to return rows from one table that have no matching records in another table )

df_anti= o.join(c, on= "customer_id", how= "anti")
display(df_anti)
```

```python
#using join with filter

df_inner= o.join(c, on= "customer_id", how= "inner").filter(o["customer_id"]==1)
display(df_inner)
```
