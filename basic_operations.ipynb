### Row Filterning

```python
df_filtered = df.filter((df.release_year >= 2000) & (df.release_year <= 2010))
display(df_filtered)
```

```python
from pyspark.sql.functions import col

df_filtered1= df.filter(col("release_year").between(2000, 2010))
display(df_filtered1)
```

```python
df_marvel= df.filter(col("studio")== "Marvel Studios")
display(df_marvel)
```

```python
df_marvel1= df.filter(df.studio=="Marvel Studios")
df_marvel1.show()
```
