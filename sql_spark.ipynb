### Creting new dataframe 

```python
from pyspark.sql import functions as f, types as T

data= [
    ("2017-01-01", 32.0, 6.0, "Rain"), 
    ("2017-01-04", None, 9.0, "Sunny"), 
    ("2017-01-05", 28.0, 10.0, "Sunny"), 
    ("2017-01-06", 24.0, 10.0, "Snow"), 
    ("2017-01-07", 20.0, 10.0, "Snow"), 
    ("2017-01-08", 16.0, 10.0, "Snow"), 
    ("2017-01-09", 12.0, 10.0, "Snow"), 
    ("2017-01-10", 8.0, 10.0, "Snow"), 
    ("2017-01-11", 4.0, 10.0, "Snow")
]

schema= "day string, temp double, humidity double, weather string"    #Defining our schema

df= spark.createDataFrame(data, schema)  #Creating DataFrame (passing argumnets like what is our data, schema)
display(df)


```

```python
from pyspark.sql.functions import col, lit

df= df.withColumn("day", f.to_date(col("day"), "yyyy-MM-dd"))   #Normalizing day column
```

```python
display(df)
```

### We want to use SQL for wether data

```python
df.createOrReplaceTempView("weather") # create a temp view
```

```python
%sql

select weather, round(avg(temp), 2) as avg_temp
from weather
group by weather 
order by avg_temp desc;
```
